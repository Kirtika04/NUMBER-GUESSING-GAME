# NUMBER GUESSING GAME

##### It is a number guessing where a number is randomly selected from a range of numbers and the players has to guess the correct number within 5 attempts

```python
#LIBRARIES
import random
```

```python
def randomnumber(n=100): #DEFAULT VALUE FOR MAX RANGE
    return random.randint(0,n)    
```

```python
#SET OF RULES
def rules():
    print("THERE ARE 5 CHANCES TO GUESS THE NUMBER")
    print("THE NUMBER RANGES FROM 0 TO 100")
    print("IN EACH ATTEMPT IF THE ANSWER IS WRONG A HINT WILL BE GIVEN")
    print("HINT WILL TELL YPOU IF THE ACTUAL NUMBER IS LESS OR GREATER THAN YOUR GUESSED VALUE")
    print("IF YOU GUESSED CORRECTLY WITHIN 5 ATTEMPTS YOU WIN!!! 😃")
    print("ELSE YOU LOSE 😭" )
```

```python
#INTRODUCTION 
def start(name):
    print("HELLO",name,", ALL THE BEST!!!")  
    print("GAME STARTS")
```

```python
#MAIN GAME
def numberguessinggame():
    key,found=randomnumber(),0
    print("NUMBER GUESSING GAME")
    name=input("ENTER YOUR NAME:")
    start(name) 
    rules()
    #CHANGE LOOP RANGE TO ANY NUMBER FOR INCREASING OR DECREASING ATTEMPTS
    for i in range(5):
        guess=int(input("ENTER YOUR GUESSED NUMBER:"))
        if guess==key:
            print("YOU WON!!!")
            found=1
            break
        elif guess>key:
            print("YOU GUESSED A BIT LARGE NUMBER!!!")
        else:
            print("YOU GUESSED A BIT SMALL NUMBER!!!")
    if found==0:
        print("YOU LOSE")
    print("THE NUMBER IS:",key)
    print("GAME ENDS")
```

```python
numberguessinggame()
```
