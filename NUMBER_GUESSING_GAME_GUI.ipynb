# NUMBER_GUESSING_GAME_GUI

##### It is a number guessing where a number is randomly selected from a range of numbers and the players has to guess the correct number within 5 attempts.This is a GUI implementation using tkinter.

```python
import random
import tkinter as tk
```

```python
def reset():
    l1.pack_forget()
    l2.pack_forget()
    l3.pack_forget()
    b1.pack_forget()
    b2.pack_forget()
    e1.pack_forget()
    e2.pack_forget()
```

```python
def sets():
    l1.pack()
    l2.pack()
    b1.pack()
```

```python
def randomnumber(n=100): #DEFAULT VALUE FOR MAX RANGE
    return random.randint(0,n)    
```

```python
def numberguessinggame():
    global attempt,key
    if attempt==0:
        key=randomnumber()
    reset()
    l2.config(text=f"HELLO {name} !!! ALL THE BEST")
    l3.config(text="ENTER YOUR GUESSED NUMBER")
    sets()
    l3.pack(after=l2)
    e2.pack(after=l3)
    b1.config(text="SUBMIT",command=check)
```

```python
def check():
    global attempt
    guessed=int(e2.get())
    attempt+=1
    if guessed==key:
        l3.config(text="YOU WON !!!")
        b1.config(text="EXIT",command=root.destroy)
    elif attempt==5:
        l3.config(text=f"YOU LOSE !!! THE NUMBER WAS {key}")
        attempt=0
        b1.config(text="PLAY AGAIN",command=numberguessinggame)
        b2.config(command=root.destroy)
        b2.pack(after=b1)
    elif guessed>key:
        l3.config(text="NUMBER GUESSED IS TOO HIGH")
    else:
        l3.config(text="NUMBER GUESSED IS TOO LOW")
```

```python
def getname():
    global name
    name=e1.get()
    numberguessinggame()
```

```python
def introduction():
    reset()
    l2.config(text="ENTER YOUR NAME:")
    b1.config(text="SUBMIT",command=getname)
    sets()
    e1.pack(after=l2)
```

```python
def start():
    reset()
    l2.config(text="THERE ARE 5 CHANCES TO GUESS THE NUMBER\n"
    "THE NUMBER RANGES FROM 0 TO 100\n"
    "IN EACH ATTEMPT IF THE ANSWER IS WRONG A HINT WILL BE GIVEN\n"
    "HINT WILL TELL YOU IF THE ACTUAL NUMBER IS LESS OR GREATER THAN YOUR GUESSED VALUE\n"
    "IF YOU GUESSED CORRECTLY WITHIN 5 ATTEMPTS YOU WIN!!! 😃\n"
    "ELSE YOU LOSE 😭")
    b1.config(text="OK",command=introduction)
    sets()
```

```python
root=tk.Tk()
root.title("NUMBER GUESSING GAME")
root.geometry("1500x500")
l1=tk.Label(root,text="NUMBER GUESSING GAME",font=("Arial",20,"bold"),pady=10)
l2=tk.Label(root,text="It is a number guessing where a number is randomly selected from a range of numbers and the players has to guess the correct number within 5 attempts",font=("Arial",10,"italic"),pady=10)
l3=tk.Label(root,text="",font=("Arial",10,"italic"),pady=10)
l4=tk.Label(root,text="",font=("Arial",10,"italic"),pady=10)
b1=tk.Button(root,text="LET'S BEGIN",bg="green",fg="white",width=20)
b2=tk.Button(root,text="EXIT",bg="green",fg="white",width=20)
e1=tk.Entry(root,width=20)
e2=tk.Entry(root,width=20)
name=""
attempt=0
key=0
b1.config(command=start)
sets()
root.mainloop()
```
